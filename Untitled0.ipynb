```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
P = np.array([100, 80, 60])

print("City A =", P[0])
print("City B =", P[1])
print("City C =", P[2])
```

```python
M = np.array([
    [0.8, 0.1, 0.1],
    [0.1, 0.8, 0.1],
    [0.1, 0.1, 0.8]
])

print(M)
```

```python
P1 = M @ P

print("After 1 year:")
print(P1)
```

```python
n = 5

Mn = np.linalg.matrix_power(M, n)

print("M^5 =")
print(Mn)
```

```python
P5 = Mn @ P

print("Population after 5 years:")
print(P5)
```

```python
values, vectors = np.linalg.eig(M)

print("Eigenvalues:")
print(values)

print("Eigenvectors:")
print(vectors)
```

```python
population = [P]

for i in range(5):
    P = M @ P
    population.append(P)

population = np.array(population)

plt.plot(population[:,0], label="City A")
plt.plot(population[:,1], label="City B")
plt.plot(population[:,2], label="City C")

plt.xlabel("Year")
plt.ylabel("Population")
plt.title("City Population Migration")
plt.legend()
plt.grid()
plt.show()
```
